```python
import scipy.stats as st
import math
n=10
# Exact 95% interval for sigma/s based on chi-square, n-1 df
lo=math.sqrt((n-1)/st.chi2.ppf(0.975,n-1))
hi=math.sqrt((n-1)/st.chi2.ppf(0.025,n-1))
# Clopper-Pearson upper 95% two-sided and one-sided for 0/10 overshoots
upper_two=st.beta.ppf(0.975,1,n)
upper_one=st.beta.ppf(0.95,1,n)
# SE of mean relative to SD
se_factor=1/math.sqrt(n)
print(f"95% CI for true SD/CV is approximately [{lo:.3f}, {hi:.3f}] × observed s/CV (normality assumption).")
print(f"SE(mean) = {se_factor:.3f} × SD.")
print(f"For 0 overshoots in 10: upper 95% CP bound = {upper_two:.3f} two-sided, {upper_one:.3f} one-sided.")
print(f"Total primary doses at 10 powders × 4 masses × 10 replicates = {10*4*10}.")
```
